```python
#load yolov8n-cls.pt model
from ultralytics import YOLO
model = YOLO("yolov8n-cls.pt")  # load a pretrained model (recommended for training)
# Train the model
path = "./solar_panel_data"

# train without data.yaml file
model.train(data=path, epochs=50, imgsz=640, batch=16)

# evaluate the model
metrics = model.val()
print(metrics)
```
